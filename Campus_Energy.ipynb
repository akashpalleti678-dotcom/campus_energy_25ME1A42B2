```python
# Program 3: A Function to calculate electricity cost
def calculate_bill(kwh,rate_per_unit):
    return kwh * rate_per_unit

today_bill = calculate_bill(600,8)
print("Today's estimated bill is Rs.", today_bill)
```

```python
# program 5:import the panda as library
import pandas as pd
df=pd.read_csv("campus_energy.csv")
print("Here is the raw data:")
display(df.head())
building_totals = df.groupby('BUILDING')['KWH_USED'].sum()
print("\ntotal usage per building:")
print(building_totals)
print("\nstatical summary:")
display(df.describe())
```
